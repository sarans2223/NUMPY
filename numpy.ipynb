```python
import numpy as np

# TASK 1: Identify Missing Values

arr = np.array([10, np.nan, 20, 30, np.nan, 40])
print("TASK 1")
print(np.isnan(arr))



# TASK 2: Count Missing Values

print("\nTASK 2")
print("Number of missing values:", np.sum(np.isnan(arr)))



# TASK 3: Replace Missing Values with 0


print("\nTASK 3")

arr_zero = np.nan_to_num(arr, nan=0)

print(arr_zero)

# TASK 4: Replace Missing Values with Mean

print("\nTASK 4")

mean = np.nanmean(arr)

arr_mean = np.where(np.isnan(arr), mean, arr)

print("Mean:", mean)
print(arr_mean)


# TASK 5: Remove Rows with Missing Values

print("\nTASK 5")

arr2 = np.array([
    [1, 2],
    [3, np.nan],
    [5, 6],
    [np.nan, np.nan]
])

arr2 = arr2[~np.isnan(arr2).any(axis=1)]

print(arr2)



# TASK 6: Interpolate Missing Values

print("\nTASK 6")

arr3 = np.array([10, np.nan, np.nan, 40, 50])

x = np.arange(len(arr3))

valid = ~np.isnan(arr3)

arr3 = np.interp(x, x[valid], arr3[valid])

print(arr3)
```
